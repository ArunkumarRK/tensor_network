**Creating tensors**

```python
scalar=tf.constant(3)
scalar, type(scalar)
```

```python
scalar.ndim
```

```python
vector=tf.constant(np.arange(10))
vector, vector.shape, vector.ndim
```

```python
tensor_n=tf.constant(range(100), shape=(5,2,2,5))

tensor_n, tensor_n.shape, tensor_n.ndim
```

```python
tensor_variable=tf.Variable([1,2,3,4],dtype=tf.float16)
tensor_variable
```

```python
tensor_variable.dtype
```

```python
tensor_variable.dtype.ndim
```

```python
tensor_variable[0].assign(5)
tensor_variable[1].assign(10)
```

**Uniform distribution**

```python
tensor_random=tf.random.uniform(shape=(2,2))
tensor_random
```

**Normal Distribution**

```python
tensor_random_normal=tf.random.normal(shape=(2,3))
tensor_random_normal
```

```python
#every run should return same random value based on specific number
tf.random.set_seed(42)
tensor_random_ss=tf.random.uniform(shape=(2,3))
tensor_random_ss  #it'll return same values again and again
```
